```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [11, 13, 10, 15, 14, 16]

plt.subplot(2,2,1)  #rows,columns, chartnum
plt.plot(x, y, marker='o')


x = [5,6,7,8,9]
y = [67,50,66,56,82]

plt.subplot(2,2,2)
plt.bar(x,y)



x = [2,4,6,8,10]
y = [57,50,60,55,20]

plt.subplot(2,2,3)
plt.step(x,y)


x = [1,3,5,7,9]
y = [67,51,62,50,23]

plt.subplot(2,2,4)
plt.scatter(x,y)
plt.show()


```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [11, 13, 10, 15, 14, 16]

plt.subplot(1,2,1)
plt.title("Age")
plt.plot(x, y, marker='o')


x = [5,6,7,8,9]
y = [67,50,66,56,82]

plt.subplot(1,2,2)
plt.title("Weight")
plt.suptitle("Employee data")
plt.bar(x,y)
```

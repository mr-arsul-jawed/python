```python
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

week1 = [10, 12, 9, 14, 13, 15, 11]
# week2 = [11, 13, 10, 15, 14, 16, 12]
# week3 = [12, 14, 11, 16, 15, 17, 13]

plt.step(days, week1, where='pre', label='Week 1', color='#FF9999')   
plt.legend(loc='upper right')
plt.title("Weekly Activity Distribution", fontsize=14)
plt.xlabel("Days of the Week", fontsize=12)
plt.ylabel("Activity Level", fontsize=12)
plt.show()
```

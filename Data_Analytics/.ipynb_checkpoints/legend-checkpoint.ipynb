```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [11, 13, 10, 15, 14, 16]
y1 = [12, 14, 11, 16, 15, 17]


plt.figure(figsize = [6,3])
plt.plot(x, y, label='Male', marker='o')
plt.plot(x, y1, label='Female', marker='s')
plt.legend(loc=2,bbox_to_anchor = (0.8, 1.2))
plt.show()
```

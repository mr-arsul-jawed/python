```python
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

week1 = [10, 12, 9, 14, 13, 15, 11]
week2 = [11, 13, 10, 15, 14, 16, 12]
week3 = [12, 14, 11, 16, 15, 17, 13]

plt.stackplot(days, week1, week2, week3, labels=['Week 1', 'Week 2', 'Week 3'], colors=['#FF9999', '#99FF99', '#9999FF'])
plt.legend(loc='upper right')
plt.title("Weekly Activity Distribution", fontsize=14)
plt.xlabel("Days of the Week", fontsize=12)
plt.ylabel("Activity Level", fontsize=12)
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

data = pd.read_excel("Data/expense.xlsx")
# print(data)


df  = pd.DataFrame(data)
plt.stackplot(df["Sub-Category"], df["Amount"], labels=['Amount'], colors=['#FFA500'])
plt.title("Expense Distribution by Sub-Category", fontsize = 13)
plt.xlabel("Sub-Category", fontsize = 12.5)
plt.ylabel("Amount Spent", fontsize = 12.5)
plt.legend()
plt.show()
```

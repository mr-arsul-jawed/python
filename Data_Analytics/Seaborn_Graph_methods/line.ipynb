```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt


data = {"Day":["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
        "Temperature":[22, 21, 23, 24, 26, 27, 25]
        }

df = pd.DataFrame(data)
print(df)
sns.lineplot(data=df, x="Day", y="Temperature", marker="o")
plt.show()

```

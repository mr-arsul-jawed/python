```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

data = sns.load_dataset("tips")
# print(data)

gp = data.groupby("day").agg({"tip":"mean"})
sns.heatmap(gp)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd


data = pd.read_excel("../Data/ESD.xlsx")
# print(data)



gp = data.groupby("Job Title").agg({"Annual_Salary":"mean"})
# print(gp)

sns.heatmap(gp)
plt.show()
```
